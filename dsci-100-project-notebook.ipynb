Visualization 1: This bar chart highlights the relationship between the number of total season wins and the proportion of teams that were able to achieve a top 8 tournament finish. This visualization is relevant to the research question because from this graph, we can see whether a higher amount of total season wins would lead to a higher chance of finishing in the top 8 based on observing the proportions. 

```r
march_madness_viz1<-march_madness_tidy|>

#We use fill here to separate the top 8 finish status (true or false) by colour
ggplot(aes(x = total_wins, fill = top_8_finish))+ 

#Here, we are using a bar graph with the position="dodge" to make a cleaner side by side visualization
geom_bar(position = "dodge") + 

#A colourblind palette is used here for easier visualization
scale_fill_brewer(palette = "Set2")+

#Human readable labels and axes are added 
labs(title ="Count of Top 8 Finishes by Total Season Wins", 
       x ="Total Season Wins",
       y = "Count of Teams",
       fill = "Top 8 Finish") 
march_madness_viz1
```

Visualization 2: This scatter plot helps examine the relationship between the conference rank and total wins with individual data points coloured to indicate whether a team finishes in the top 8 tournament finish. From this visualization, it allows us to assess how a team's ranking in the conference relates to the total wins and the chances they'll qualify into the top 8 in the tournament. 

```r
march_madness_viz2<- march_madness_tidy |> 

#Filtering here for any NA values in the conf_rank column
filter(!is.na(conf_rank))|> 

#Setting up ggplot comparing conf_rank and total_wins, then separating by colour
ggplot(aes(x = factor(conf_rank), y = total_wins, color = top_8_finish)) + 

#Creating scatterpoint graph with slight transparency (using alpha) and custom point size for better visualization
geom_point(alpha = 0.8,size=4) + 

#Using colourblind friendly colour palette
scale_color_brewer(palette = "Dark2") + 

#Adding human readable labels and axes
labs(title = "Total Wins vs. Conference Rank",
       x = "Conference Rank (1=Best)", 
       y = "Total Wins", 
       color = "Top 8 Finish")
march_madness_viz2
```
